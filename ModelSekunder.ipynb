```python
import google.generativeai as genai


def get_agroscan_advice(plant_name, disease_name):
    """
    Fungsi ini memanggil Gemini API (Model Sekunder) untuk memberikan saran pertanian
    menggunakan System Instructions agar output konsisten dan terstruktur.
    """

    # --- 1. SYSTEM INSTRUCTION ---
    system_prompts = """
    Kamu adalah AgroScan, konsultan ahli pertanian dan patologi tanaman yang cerdas.
    Tugasmu adalah memberikan saran penanganan penyakit tanaman kepada petani.
    
    Setiap jawaban WAJIB memuat 4 poin berikut secara LENGKAP dan TERSTRUKTUR:
    1. PENJELASAN SINGKAT: Apa itu penyakit ini dan ciri fisiknya.
    2. PENYEBAB: Apa pemicu utamanya (Jamur/Virus/Bakteri/Lingkungan).
    3. PENANGANAN (SOLUSI):
       - Cara Organik/Alami (Ramah lingkungan).
       - Cara Kimiawi (Rekomendasi bahan aktif fungisida/pestisida).
    4. PENCEGAHAN: Tips agar tanaman tidak terserang lagi kedepannya.
    
    Jawablah dengan Bahasa Indonesia yang jelas, tegas, dan mudah dipahami petani.
    PENTING: JANGAN memotong jawaban. Pastikan poin 4 selalu tertulis sampai selesai.
    """

    # --- 2. KONFIGURASI GENERASI ---
    config = genai.types.GenerationConfig(
        max_output_tokens=2048, 
        temperature=0.7
    )

    # --- 3. INISIALISASI MODEL ---
    try:
        model = genai.GenerativeModel(
            model_name='gemini-flash-latest',
            system_instruction=system_prompts, 
            generation_config=config
        )

        # --- 4. INPUT USER ---
        user_prompt = f"Tanaman: {plant_name}, Penyakit: {disease_name}"

        # Eksekusi
        response = model.generate_content(user_prompt)
        return response.text

    except Exception as e:
        return f"⚠️ Maaf, terjadi kesalahan koneksi ke server AI: {e}"
```

```python
# --- SIMULASI INTEGRASI ---
test_plant = "Tomat (Tomato)"
test_disease = "Early Blight (Bercak Daun Awal)"

print(
    f"Sedang meminta saran untuk {test_plant} dengan penyakit {test_disease}...\n")

# Panggil Model Sekunder
saran_agroscan = get_agroscan_advice(test_plant, test_disease)

# Tampilkan Output
print("="*50)
print("HASIL ANALISIS AGROSCAN (MODEL SEKUNDER)")
print("="*50)
print(saran_agroscan)
```
